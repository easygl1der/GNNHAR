## 构建邻接矩阵

```python
import numpy as np
import pandas as pd
from sklearn.covariance import GraphicalLassoCV

def GLASSO_Precision(subret):
    """
    使用Graphical Lasso方法计算精度矩阵并构建邻接矩阵

    Args:
        subret: 股票收益率数据框，每列代表一只股票的收益率时间序列

    Returns:
        adj_df: 标准化后的邻接矩阵，用于表示股票间的关系网络
    """

    # 获取股票数量和股票代码
    n = subret.shape[1]
    tickers = subret.columns

    # 使用交叉验证的Graphical Lasso拟合数据，自动选择最优的正则化参数alpha
    cov = GraphicalLassoCV().fit(subret)
    print('Alpha in GLASSO: %.3f' % cov.alpha_)

    # 从精度矩阵中提取非零元素，构建二值化的相关性矩阵
    corr = cov.precision_ != 0
    print('Sparsity of Adj: %.3f' % corr.mean())

    # 去除对角线元素（自相关），得到邻接矩阵
    corr_adj = corr - np.identity(n)

    # 计算度矩阵的逆平方根，用于标准化邻接矩阵
    # 加上小常数1e-8防止除零错误
    d_sqrt_inv = np.diag(np.sqrt(1/(corr_adj.sum(1)+1e-8)))

    # 应用对称标准化：D^(-1/2) * A * D^(-1/2)
    # 这样可以确保邻接矩阵的特征值在合理范围内
    adj_df = pd.DataFrame(np.dot(np.dot(d_sqrt_inv, corr_adj), d_sqrt_inv),
                         columns=tickers, index=tickers)

    return adj_df

print("GLASSO_Precision function defined.")



# Prepare the data for GLASSO_Precision by dropping the 'Date' column
subret_data = dow30_df.drop(columns=['Date'])

# Calculate the adjacency matrix
adjacency_matrix = GLASSO_Precision(subret_data)

# Display the first few rows and columns of the adjacency matrix
print('\nAdjacency Matrix (first 5 rows and columns):')
display(adjacency_matrix)
```

## Data Preprocessing for HAR Models

Prepare data for HAR-based volatility forecasting. The HAR model uses three components:
- **Daily**: 1-day lagged volatility
- **Weekly**: 5-day average lagged volatility  
- **Monthly**: 22-day average lagged volatility

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# Set random seeds for reproducibility
np.random.seed(42)
torch.manual_seed(42)

print("Libraries imported successfully!")

# Prepare RV data (target variable)
rv_data = rv_df.set_index('Date')
print(f"RV data shape: {rv_data.shape}")
print(f"Date range: {rv_data.index[0]} to {rv_data.index[-1]}")

# Prepare return data (for adjacency matrix)
ret_data = dow30_df.set_index('Date')
print(f"Return data shape: {ret_data.shape}")
print(f"Date range: {ret_data.index[0]} to {ret_data.index[-1]}")
```

```python
def create_HAR_features(rv_data, lags=[1, 5, 22]):
    """
    Create HAR (Heterogeneous Autoregressive) features
    
    Args:
        rv_data: DataFrame with RV values (Date index, stocks as columns)
        lags: List of lag periods [daily, weekly, monthly]
    
    Returns:
        DataFrame with HAR features for each stock
    """
    print(f"Input rv_data shape: {rv_data.shape}")
    print(f"Input rv_data columns: {rv_data.columns.tolist()}")
    print(f"Input rv_data index type: {type(rv_data.index)}")
    
    feature_list = []
    
    for ticker in rv_data.columns:
        print(f"Processing ticker: {ticker}")
        ticker_series = rv_data[ticker].copy()
        
        # Create DataFrame for this ticker
        ticker_df = pd.DataFrame()
        ticker_df['Date'] = rv_data.index
        ticker_df['Ticker'] = ticker
        ticker_df['Target'] = ticker_series.values
        
        # Create lagged features
        for lag in lags:
            # Calculate rolling mean of past 'lag' periods
            lag_feature = ticker_series.rolling(window=lag, min_periods=1).mean().shift(1)
            ticker_df[f'lag_{lag}'] = lag_feature.values
        
        # Remove rows with NaN values (first few rows due to lags)
        ticker_df_clean = ticker_df.dropna()
        print(f"  {ticker}: {len(ticker_df_clean)} valid samples after removing NaN")
        
        if len(ticker_df_clean) > 0:
            feature_list.append(ticker_df_clean)
    
    if len(feature_list) == 0:
        print("ERROR: No valid data after processing!")
        return pd.DataFrame()
    
    # Combine all tickers
    df = pd.concat(feature_list, ignore_index=True)
    print(f"Combined data shape: {df.shape}")
    
    return df

# Create HAR features
print("Creating HAR features...")
har_data = create_HAR_features(rv_data)
print(f"HAR data shape: {har_data.shape}")

if len(har_data) > 0:
    print("\nFirst few rows:")
    display(har_data.head())
    print(f"\nFeature columns: {[col for col in har_data.columns if 'lag' in col]}")
else:
    print("ERROR: HAR data is empty!")
    print("Checking original data...")
    print(f"rv_data shape: {rv_data.shape}")
    print(f"rv_data head:")
    display(rv_data.head())
```

## Train-Test Split

We'll use a time-based split:
- **Training**: First 80% of the data
- **Testing**: Last 20% of the data

```python
# Get unique dates
dates = sorted(har_data['Date'].unique())
print(f"Total dates: {len(dates)}")
print(f"Sample dates: {dates[:5] if len(dates) > 0 else 'No dates found'}")

# Check if we have data
if len(dates) == 0:
    print("ERROR: No dates found in har_data!")
    print(f"har_data shape: {har_data.shape}")
    print(f"har_data columns: {har_data.columns.tolist()}")
    display(har_data.head())
else:
    # 80-20 split
    split_idx = int(len(dates) * 0.8)
    if split_idx == 0:
        split_idx = max(1, len(dates) - 1)  # Ensure at least 1 training sample
    
    train_dates = dates[:split_idx]
    test_dates = dates[split_idx:]
    
    print(f"Training dates: {len(train_dates)} ({train_dates[0]} to {train_dates[-1]})")
    print(f"Testing dates: {len(test_dates)} ({test_dates[0]} to {test_dates[-1]})")
    
    # Split data
    train_data = har_data[har_data['Date'].isin(train_dates)].copy()
    test_data = har_data[har_data['Date'].isin(test_dates)].copy()
    
    print(f"\nTraining samples: {len(train_data)}")
    print(f"Testing samples: {len(test_data)}")
    
    # Prepare feature columns
    feature_cols = [col for col in har_data.columns if 'lag' in col]
    print(f"\nFeature columns: {feature_cols}")
```

## Model 1: HAR (Baseline Model)

The standard HAR model without graph structure. Each stock is modeled independently using its own lagged volatilities.

```python
# Train HAR model (Linear Regression)
print("Training HAR model...")

X_train = train_data[feature_cols].values
y_train = train_data['Target'].values

X_test = test_data[feature_cols].values
y_test = test_data['Target'].values

# Fit linear regression
har_model = LinearRegression()
har_model.fit(X_train, y_train)

# Make predictions
y_train_pred_har = har_model.predict(X_train)
y_test_pred_har = har_model.predict(X_test)

# Handle negative predictions (volatility must be positive)
y_train_pred_har = np.maximum(y_train_pred_har, 0.01)
y_test_pred_har = np.maximum(y_test_pred_har, 0.01)

print("HAR model training completed!")
print(f"Coefficients: {har_model.coef_}")
print(f"Intercept: {har_model.intercept_}")
```

## Model 2: GHAR (Graph HAR with GLASSO Adjacency)

GHAR extends HAR by incorporating cross-sectional dependencies through the GLASSO adjacency matrix. The model applies graph convolution to capture relationships between stocks.

```python
def apply_graph_convolution(df, adj_matrix, feature_cols):
    """
    Apply graph convolution: multiply adjacency matrix with features
    
    Args:
        df: DataFrame with features
        adj_matrix: Adjacency matrix (stocks x stocks)
        feature_cols: List of feature column names
    
    Returns:
        DataFrame with graph-convolved features
    """
    result_list = []
    
    for date in df['Date'].unique():
        date_df = df[df['Date'] == date].copy()
        
        # Get features for all stocks at this date (N x F matrix)
        features = date_df[feature_cols].values
        
        # Apply graph convolution: A @ X  (adjacency @ features)
        conv_features = adj_matrix.values @ features
        
        # Create new dataframe with convolved features
        conv_df = pd.DataFrame(
            conv_features,
            columns=[f'gc_{col}' for col in feature_cols]
        )
        conv_df['Date'] = date
        conv_df['Ticker'] = date_df['Ticker'].values
        conv_df['Target'] = date_df['Target'].values
        
        # Concatenate original and convolved features
        for col in feature_cols:
            conv_df[col] = date_df[col].values
        
        result_list.append(conv_df)
    
    return pd.concat(result_list, ignore_index=True)

# Apply graph convolution to training and test data
print("Applying graph convolution with GLASSO adjacency matrix...")
train_data_ghar = apply_graph_convolution(train_data, adjacency_matrix, feature_cols)
test_data_ghar = apply_graph_convolution(test_data, adjacency_matrix, feature_cols)

print(f"GHAR training data shape: {train_data_ghar.shape}")
print(f"GHAR testing data shape: {test_data_ghar.shape}")

# New feature columns include both original and graph-convolved features
ghar_feature_cols = [col for col in train_data_ghar.columns if 'lag' in col or 'gc_' in col]
print(f"\nGHAR feature columns ({len(ghar_feature_cols)}): {ghar_feature_cols}")
```

```python
# Train GHAR model
print("Training GHAR model...")

X_train_ghar = train_data_ghar[ghar_feature_cols].values
y_train_ghar = train_data_ghar['Target'].values

X_test_ghar = test_data_ghar[ghar_feature_cols].values
y_test_ghar = test_data_ghar['Target'].values

# Fit linear regression with graph features
ghar_model = LinearRegression()
ghar_model.fit(X_train_ghar, y_train_ghar)

# Make predictions
y_train_pred_ghar = ghar_model.predict(X_train_ghar)
y_test_pred_ghar = ghar_model.predict(X_test_ghar)

# Handle negative predictions
y_train_pred_ghar = np.maximum(y_train_pred_ghar, 0.01)
y_test_pred_ghar = np.maximum(y_test_pred_ghar, 0.01)

print("GHAR model training completed!")
print(f"Number of coefficients: {len(ghar_model.coef_)}")
print(f"Intercept: {ghar_model.intercept_:.4f}")
```

## Model 3: GNNHAR (Graph Neural Network HAR)

GNNHAR uses neural networks with graph convolutional layers to model non-linear relationships in volatility forecasting.

```python
# Define Graph Convolutional Layer
class GraphConvLayer(nn.Module):
    """Graph Convolutional Layer: H' = A @ H @ W + b"""
    def __init__(self, in_features, out_features, bias=True):
        super(GraphConvLayer, self).__init__()
        self.weight = nn.Parameter(torch.FloatTensor(in_features, out_features))
        nn.init.xavier_uniform_(self.weight, gain=nn.init.calculate_gain('relu'))
        
        if bias:
            self.bias = nn.Parameter(torch.FloatTensor(1, out_features))
            nn.init.ones_(self.bias)
        else:
            self.bias = None
    
    def forward(self, node_feature, adj):
        """
        Args:
            node_feature: [batch_size, N, in_features]
            adj: [N, N] adjacency matrix
        Returns:
            [batch_size, N, out_features]
        """
        h = torch.matmul(node_feature, self.weight)  # H @ W
        output = torch.matmul(adj, h)  # A @ (H @ W)
        if self.bias is not None:
            return output + self.bias
        return output


# Define GNNHAR1L Model (1-layer GNN)
class GNNHAR1L(nn.Module):
    """
    GNNHAR with 1 graph convolutional layer
    Architecture: Linear HAR branch + GCN branch with residual connection
    """
    def __init__(self, n_hid=16):
        super(GNNHAR1L, self).__init__()
        
        # Linear HAR component
        self.linear1 = nn.Linear(3, 1, bias=True)
        
        # Graph neural network component
        self.gcn1 = GraphConvLayer(3, n_hid, bias=False)
        self.mlp1 = nn.Linear(n_hid, 1, bias=False)
        self.relu = nn.ReLU()
    
    def forward(self, node_feat, adj):
        """
        Args:
            node_feat: [batch_size, N, 3] - HAR features (daily, weekly, monthly)
            adj: [N, N] - adjacency matrix
        Returns:
            [batch_size, N] - predicted volatilities
        """
        # Linear HAR branch
        H1 = self.linear1(node_feat)  # [batch_size, N, 1]
        
        # GNN branch
        H2 = self.gcn1(node_feat, adj)  # [batch_size, N, n_hid]
        H2 = self.relu(H2)
        H2 = self.mlp1(H2)  # [batch_size, N, 1]
        
        # Residual connection
        res = H1 + H2
        res = self.relu(res)  # Ensure positive volatility
        
        return res.squeeze(-1)  # [batch_size, N]

print("GNNHAR model classes defined!")
```

```python
# Prepare data for GNNHAR (batch format)
def prepare_gnnhar_data(df, feature_cols, n_stocks):
    """
    Prepare data in batch format for GNNHAR
    Each sample is all stocks at a single date
    
    Args:
        df: DataFrame with features
        feature_cols: List of feature columns
        n_stocks: Number of stocks
    
    Returns:
        X: [n_dates, n_stocks, n_features]
        y: [n_dates, n_stocks]
    """
    dates = sorted(df['Date'].unique())
    X_list = []
    y_list = []
    
    for date in dates:
        date_df = df[df['Date'] == date].sort_values('Ticker')
        if len(date_df) == n_stocks:  # Ensure we have all stocks
            X_list.append(date_df[feature_cols].values)
            y_list.append(date_df['Target'].values)
    
    X = np.array(X_list)  # [n_dates, n_stocks, n_features]
    y = np.array(y_list)  # [n_dates, n_stocks]
    
    return X, y

# Get number of stocks
n_stocks = len(rv_data.columns)
print(f"Number of stocks: {n_stocks}")

# Prepare data
X_train_gnn, y_train_gnn = prepare_gnnhar_data(train_data, feature_cols, n_stocks)
X_test_gnn, y_test_gnn = prepare_gnnhar_data(test_data, feature_cols, n_stocks)

print(f"X_train_gnn shape: {X_train_gnn.shape}")  # [n_dates, n_stocks, 3]
print(f"y_train_gnn shape: {y_train_gnn.shape}")  # [n_dates, n_stocks]
print(f"X_test_gnn shape: {X_test_gnn.shape}")
print(f"y_test_gnn shape: {y_test_gnn.shape}")

# Convert to PyTorch tensors
X_train_tensor = torch.FloatTensor(X_train_gnn)
y_train_tensor = torch.FloatTensor(y_train_gnn)
X_test_tensor = torch.FloatTensor(X_test_gnn)
y_test_tensor = torch.FloatTensor(y_test_gnn)

# Convert adjacency matrix to tensor
adj_tensor = torch.FloatTensor(adjacency_matrix.values)

print("\nData conversion completed!")
```

```python
# Initialize model
n_hid = 16
gnnhar_model = GNNHAR1L(n_hid=n_hid)
print(f"Model initialized with {n_hid} hidden units")
print(gnnhar_model)

# Training parameters
n_epochs = 500
learning_rate = 0.001
batch_size = 32

# Loss function and optimizer
criterion = nn.MSELoss()
optimizer = optim.Adam(gnnhar_model.parameters(), lr=learning_rate)

# Create DataLoader
train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)

print(f"\nTraining configuration:")
print(f"  Epochs: {n_epochs}")
print(f"  Learning rate: {learning_rate}")
print(f"  Batch size: {batch_size}")
print(f"  Number of batches: {len(train_loader)}")
```

```python
# Training loop
print("Training GNNHAR model...")
train_losses = []
test_losses = []

for epoch in range(n_epochs):
    gnnhar_model.train()
    epoch_loss = 0
    
    for batch_X, batch_y in train_loader:
        # Forward pass
        predictions = gnnhar_model(batch_X, adj_tensor)
        loss = criterion(predictions, batch_y)
        
        # Backward pass
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        epoch_loss += loss.item()
    
    # Calculate average training loss
    avg_train_loss = epoch_loss / len(train_loader)
    train_losses.append(avg_train_loss)
    
    # Evaluate on test set
    gnnhar_model.eval()
    with torch.no_grad():
        test_pred = gnnhar_model(X_test_tensor, adj_tensor)
        test_loss = criterion(test_pred, y_test_tensor).item()
        test_losses.append(test_loss)
    
    # Print progress every 50 epochs
    if (epoch + 1) % 50 == 0:
        print(f"Epoch [{epoch+1}/{n_epochs}] - Train Loss: {avg_train_loss:.4f}, Test Loss: {test_loss:.4f}")

print("\nTraining completed!")
```

```python
# Plot training curves
plt.figure(figsize=(10, 5))
plt.plot(train_losses, label='Training Loss', alpha=0.7)
plt.plot(test_losses, label='Test Loss', alpha=0.7)
plt.xlabel('Epoch')
plt.ylabel('MSE Loss')
plt.title('GNNHAR Training Progress')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Final training loss: {train_losses[-1]:.4f}")
print(f"Final test loss: {test_losses[-1]:.4f}")
```

## Error Analysis: MSE and QLIKE

We evaluate all models using two metrics:

1. **MSE (Mean Squared Error)**: $\text{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$

2. **QLIKE (Quasi-Likelihood)**: $\text{QLIKE} = \frac{1}{n}\sum_{i=1}^{n}\left(\frac{y_i}{\hat{y}_i} - \log\frac{y_i}{\hat{y}_i} - 1\right)$

QLIKE is specifically designed for volatility forecasting and is more robust to outliers than MSE.

```python
def calculate_mse(y_true, y_pred):
    """Calculate Mean Squared Error"""
    return np.mean((y_true - y_pred) ** 2)

def calculate_qlike(y_true, y_pred, epsilon=1e-4):
    """
    Calculate Quasi-Likelihood (QLIKE) loss for volatility forecasting
    QLIKE = mean(y_true/y_pred - log(y_true/y_pred) - 1)
    
    Args:
        y_true: True volatility values
        y_pred: Predicted volatility values
        epsilon: Small constant to avoid division by zero
    """
    # Ensure positive values and avoid division by zero
    y_pred = np.maximum(y_pred, epsilon)
    y_true = np.maximum(y_true, epsilon)
    
    # QLIKE formula
    ratio = y_true / y_pred
    qlike = np.mean(ratio - np.log(ratio) - 1)
    
    return qlike

print("Error metric functions defined!")
```

```python
# Calculate errors for all models
results = {
    'Model': [],
    'Train_MSE': [],
    'Test_MSE': [],
    'Train_QLIKE': [],
    'Test_QLIKE': []
}

# HAR Model
results['Model'].append('HAR')
results['Train_MSE'].append(calculate_mse(y_train, y_train_pred_har))
results['Test_MSE'].append(calculate_mse(y_test, y_test_pred_har))
results['Train_QLIKE'].append(calculate_qlike(y_train, y_train_pred_har))
results['Test_QLIKE'].append(calculate_qlike(y_test, y_test_pred_har))

# GHAR Model
results['Model'].append('GHAR')
results['Train_MSE'].append(calculate_mse(y_train_ghar, y_train_pred_ghar))
results['Test_MSE'].append(calculate_mse(y_test_ghar, y_test_pred_ghar))
results['Train_QLIKE'].append(calculate_qlike(y_train_ghar, y_train_pred_ghar))
results['Test_QLIKE'].append(calculate_qlike(y_test_ghar, y_test_pred_ghar))

# GNNHAR Model
results['Model'].append('GNNHAR1L')
results['Train_MSE'].append(calculate_mse(y_train_gnn_flat, y_train_pred_gnn_flat))
results['Test_MSE'].append(calculate_mse(y_test_gnn_flat, y_test_pred_gnn_flat))
results['Train_QLIKE'].append(calculate_qlike(y_train_gnn_flat, y_train_pred_gnn_flat))
results['Test_QLIKE'].append(calculate_qlike(y_test_gnn_flat, y_test_pred_gnn_flat))

# Create results DataFrame
results_df = pd.DataFrame(results)
print("\n" + "="*80)
print("MODEL COMPARISON - ERROR ANALYSIS")
print("="*80)
display(results_df)

# Calculate improvement over HAR baseline
print("\n" + "="*80)
print("IMPROVEMENT OVER HAR BASELINE (%)")
print("="*80)
har_test_mse = results_df[results_df['Model'] == 'HAR']['Test_MSE'].values[0]
har_test_qlike = results_df[results_df['Model'] == 'HAR']['Test_QLIKE'].values[0]

for idx, row in results_df.iterrows():
    if row['Model'] != 'HAR':
        mse_improve = (har_test_mse - row['Test_MSE']) / har_test_mse * 100
        qlike_improve = (har_test_qlike - row['Test_QLIKE']) / har_test_qlike * 100
        print(f"{row['Model']:12s}: MSE {mse_improve:+.2f}%  |  QLIKE {qlike_improve:+.2f}%")
```

## Summary and Conclusions

### Model Comparison Summary

This notebook implements and compares three volatility forecasting models:

1. **HAR (Baseline)**: Standard Heterogeneous Autoregressive model
   - Uses only temporal dependencies (daily, weekly, monthly lags)
   - Linear model trained independently for each stock

2. **GHAR (Graph HAR)**: HAR with graph structure
   - Incorporates cross-sectional dependencies via GLASSO adjacency matrix
   - Applies graph convolution to capture inter-stock relationships
   - Still a linear model but with enhanced features

3. **GNNHAR1L (Graph Neural Network HAR)**: Deep learning approach
   - 1-layer graph convolutional neural network
   - Captures non-linear relationships in volatility dynamics
   - Residual connection combines linear HAR and GCN components

### Key Findings

The adjacency matrix constructed using Graphical Lasso (GLASSO):
- **Alpha = 0.000**: Minimal regularization needed
- **Sparsity = 0.367**: ~37% of possible edges exist
- This indicates moderate connectivity in the Dow 30 stock network

### Evaluation Metrics

- **MSE (Mean Squared Error)**: Standard metric for regression
- **QLIKE (Quasi-Likelihood)**: Specifically designed for volatility forecasting, more robust to outliers

### Advantages of Graph-Based Models

1. **Cross-sectional information**: Capture relationships between stocks
2. **Network effects**: Model how volatility spillovers across the market
3. **Improved predictions**: Especially for stocks with strong inter-dependencies
