# GNNHAR Colab: GitHub + Google Drive Workflow

This notebook mounts Google Drive, clones the project branch from GitHub, and writes outputs to Drive so Colab runtime restarts do not lose results.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import os, pathlib, subprocess, textwrap

REPO_URL = 'https://github.com/easygl1der/GNNHAR.git'
BRANCH = '2026-06-01'
PROJECT_DIR = pathlib.Path('/content/GNNHAR')
DRIVE_ROOT = pathlib.Path('/content/drive/MyDrive/GNNHAR-colab-runs')
DATA_DIR = DRIVE_ROOT / 'data'
OUTPUT_DIR = DRIVE_ROOT / 'outputs'

for path in [DRIVE_ROOT, DATA_DIR, OUTPUT_DIR]:
    path.mkdir(parents=True, exist_ok=True)

if PROJECT_DIR.exists():
    subprocess.run(['rm', '-rf', str(PROJECT_DIR)], check=True)

subprocess.run([
    'git', 'clone', '--depth', '1', '--branch', BRANCH, REPO_URL, str(PROJECT_DIR)
], check=True)

print('repo cloned:', PROJECT_DIR)
print('drive root:', DRIVE_ROOT)
print('data dir:', DATA_DIR)
print('output dir:', OUTPUT_DIR)
```

```python
import json, platform, sys, pathlib, os

manifest = {
    'status': 'ok',
    'python': sys.version.split()[0],
    'platform': platform.platform(),
    'project_dir_exists': pathlib.Path('/content/GNNHAR').exists(),
    'drive_output_dir': str(OUTPUT_DIR),
}

manifest_path = OUTPUT_DIR / 'browseros_drive_workflow_test.json'
manifest_path.write_text(json.dumps(manifest, indent=2), encoding='utf-8')
print(manifest_path)
print(json.dumps(manifest, indent=2))
```
